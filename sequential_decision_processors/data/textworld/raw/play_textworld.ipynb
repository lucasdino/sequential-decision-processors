```python
import random
import textworld
import utils as util


# Load our various validation seeds
with open("utils/tw_valid_seeds/twx_valid.txt") as f:
    twx_valid_seeds = [int(x) for x in f.read().split()]

with open("utils/tw_valid_seeds/alfworld_valid_seen.txt") as f:
    alfworld_valid_seen_seeds = [x for x in f.read().split()]

with open("utils/tw_valid_seeds/alfworld_valid_unseen.txt") as f:
    alfworld_valid_unseen_seeds = [x for x in f.read().split()]
```

```python
# Playable twx
play_twx = True

# Start by resetting
if play_twx:
    seeds=random.sample(twx_valid_seeds, 1)
    env = util.TWXBatchGym(seeds, split="train", max_steps=50)

    obs, info = env.reset(seeds[0])
    print(info['admissible_commands'])
    print(f"{'='*60}\nStep 0:\n{obs[0]}\n\n")

    done, step = False, 1
    while not done:
        if play_twx:
            command = input("> ").strip()
        else:
            command = None
        if command == "stop":
            break
        print(f"ACTION: {command}")
        obs, reward, done, info = env.step([command])
        print(info['admissible_commands'])
        print(f"{'='*60}\nStep {step}:\n{obs[0]}\n\n")
        print(info)
        step += 1
        done = done[0]
```

```python
# Playable alfworld
play_alfworld = True


if play_alfworld:
    seeds=random.sample(alfworld_valid_seen_seeds, 1)
    wrappers = [util.AlfredDemangler()]
    request_infos = textworld.EnvInfos(
        won=True, lost=True, admissible_commands=True, verbs=True,
        extras=["gamefile", "walkthrough"]
    )
    env = util.AlfworldResetEnv(
        seeds,
        request_infos,
        max_episode_steps=50,
        wrappers=wrappers,
        batch_size=1,
        asynchronous=True,
    )

    # Start by resetting
    obs, info = env.reset(seeds[0])
    print(f"{'='*60}\nStep 0:\n{obs[0]}\n\n")

    done, step = False, 1 
    while not done:
        if play_alfworld:
            command = input("> ")
        else:
            command = None
        if command == "stop":
            break
        obs, reward, done, info = env.step([command])
        print(f"{'='*60}\nStep {step}:\n{obs[0]}\n\n")
        print(info)
        step += 1
        done = done[0]
```
